```python

# =====================================================
# ONLINE RETAIL CUSTOMER PURCHASE ANALYSIS SYSTEM
# Using Python, Pandas and NumPy
# =====================================================

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np

from google.colab import files

print("Libraries Imported Successfully!")

# =====================================================
# STEP 2: UPLOAD DATASET
# =====================================================

print("\nPlease Upload Your CSV Dataset")

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv(filename)

print("\nDataset Uploaded Successfully!")

# =====================================================
# STEP 3: DISPLAY DATASET
# =====================================================

print("\n========== DATASET INFORMATION ==========")

print("\nFirst 10 Records:")
display(df.head(10))

print("\nTotal Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

# =====================================================
# STEP 4: DATA CLEANING
# =====================================================

print("\n========== DATA CLEANING ==========")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

df = df.drop_duplicates()

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("\nData Cleaning Completed Successfully!")

# =====================================================
# STEP 5: DATASET SUMMARY
# =====================================================

print("\n========== DATASET SUMMARY ==========")

df.info()

print("\nStatistical Summary:")
display(df.describe())

# =====================================================
# STEP 6: SALES ANALYSIS USING NUMPY
# =====================================================

total_sales = np.sum(df["Total_Amount_INR"])

average_sales = np.mean(df["Total_Amount_INR"])

total_quantity = np.sum(df["Quantity"])

total_orders = df["Order_ID"].nunique()

total_customers = df["Customer_ID"].nunique()

print("\n========== SALES REPORT ==========")

print("Total Sales: Rs.", round(total_sales, 2))

print("Average Order Value: Rs.", round(average_sales, 2))

print("Total Quantity Sold:", total_quantity)

print("Total Orders:", total_orders)

print("Total Customers:", total_customers)

# =====================================================
# STEP 7: CATEGORY WISE SALES
# =====================================================

category_sales = df.groupby("Category")["Total_Amount_INR"].sum()

print("\n========== CATEGORY WISE SALES ==========")

display(category_sales.sort_values(ascending=False))

# =====================================================
# STEP 8: BEST SELLING PRODUCTS
# =====================================================

best_products = df.groupby("Product_Name")["Quantity"].sum()

print("\n========== TOP 10 BEST SELLING PRODUCTS ==========")

display(best_products.sort_values(ascending=False).head(10))

# =====================================================
# STEP 9: CITY WISE SALES
# =====================================================

city_sales = df.groupby("City")["Total_Amount_INR"].sum()

print("\n========== CITY WISE SALES ==========")

display(city_sales.sort_values(ascending=False))

# =====================================================
# STEP 10: PAYMENT METHOD ANALYSIS
# =====================================================

payment_analysis = df["Payment_Method"].value_counts()

print("\n========== PAYMENT METHOD ANALYSIS ==========")

display(payment_analysis)

# =====================================================
# STEP 11: MONTHLY SALES ANALYSIS
# =====================================================

df["Month"] = df["Order_Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Month")["Total_Amount_INR"].sum()

print("\n========== MONTHLY SALES REPORT ==========")

display(monthly_sales)

# =====================================================
# STEP 12: SALES CHANNEL ANALYSIS
# =====================================================

channel_analysis = df["Sales_Channel"].value_counts()

print("\n========== SALES CHANNEL ANALYSIS ==========")

display(channel_analysis)

# =====================================================
# STEP 13: ORDER STATUS ANALYSIS
# =====================================================

status_analysis = df["Order_Status"].value_counts()

print("\n========== ORDER STATUS ANALYSIS ==========")

display(status_analysis)

# =====================================================
# STEP 14: CUSTOMER RATING
# =====================================================

average_rating = np.mean(df["Customer_Rating"])

print("\n========== CUSTOMER RATING ==========")

print("Average Customer Rating:",
      round(average_rating, 2))

# =====================================================
# STEP 15: SAVE PROJECT REPORTS
# =====================================================

df.to_csv("cleaned_retail_dataset.csv", index=False)

category_sales.to_csv("category_sales_report.csv")

city_sales.to_csv("city_sales_report.csv")

monthly_sales.to_csv("monthly_sales_report.csv")

payment_analysis.to_csv("payment_methods_report.csv")

best_products.to_csv("best_selling_products_report.csv")

print("\nAll Reports Generated Successfully!")

# =====================================================
# STEP 16: PROJECT COMPLETED
# =====================================================

print("\n" + "=" * 55)

print("ONLINE RETAIL CUSTOMER PURCHASE ANALYSIS")

print("PROJECT COMPLETED SUCCESSFULLY!")

print("=" * 55)

# =====================================================
# STEP 17: DOWNLOAD REPORTS
# =====================================================

print("\nDownloading Cleaned Dataset...")

files.download("cleaned_retail_dataset.csv")

print("\nDownloading Category Sales Report...")

files.download("category_sales_report.csv")

print("\nDownloading City Sales Report...")

files.download("city_sales_report.csv")

print("\nDownloading Monthly Sales Report...")

files.download("monthly_sales_report.csv")

print("\nDownloading Payment Methods Report...")

files.download("payment_methods_report.csv")

print("\nDownloading Best Selling Products Report...")

files.download("best_selling_products_report.csv")


```
